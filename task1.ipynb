# Task 1 — Inconsistencies and Duplicates

This notebook cleans the `club_signups.csv` dataset using pandas.

The goal is to produce one clean row per student per club while keeping the original CSV unchanged.


```python
import pandas as pd
```

```python
raw = pd.read_csv("club_signups.csv")
df = raw.copy()
raw.head()
```

## Part A — Inconsistencies

### A1. Inspect the raw data

```python
print("Shape:", raw.shape)
```

```python
raw.dtypes
```

```python
categorical_columns = [
    "signup_id",
    "student_id",
    "full_name",
    "email",
    "faculty",
    "city",
    "club",
    "fee_paid"
]

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(raw[column].value_counts(dropna=False))
```

### A2. Clean faculty, club, and city

```python
for column in ["faculty", "club", "city"]:
    df[column] = df[column].str.strip().str.lower()
```

```python
for column in ["faculty", "club", "city"]:
    print(f"\n--- {column} ---")
    print(df[column].value_counts(dropna=False))
```

```python
faculty_map = {
    "met": "MET",
    "media engineering and technology": "MET",
    "iet": "IET",
    "information engineering and technology": "IET",
    "ems": "EMS",
    "engineering and materials science": "EMS",
    "pharmacy": "Pharmacy",
    "pharma": "Pharmacy",
    "management": "Management"
}

club_map = {
    "robotics": "Robotics",
    "debate": "Debate",
    "debating": "Debate",
    "debate club": "Debate",
    "music": "Music",
    "football": "Football",
    "soccer": "Football",
    "chess": "Chess",
    "chess club": "Chess"
}

city_map = {
    "cairo": "Cairo",
    "giza": "Giza",
    "el giza": "Giza",
    "alexandria": "Alexandria",
    "alex": "Alexandria"
}
df["faculty"] = df["faculty"].map(faculty_map)
df["club"] = df["club"].map(club_map)
df["city"] = df["city"].map(city_map)
```

```python
for column in ["faculty", "club", "city"]:
    print(f"\n--- {column} ---")
    print(df[column].value_counts(dropna=False))
```

```python
assert set(df["faculty"].dropna().unique()) <= {
    "MET", "IET", "EMS", "Pharmacy", "Management"
}

assert set(df["club"].dropna().unique()) <= {
    "Robotics", "Debate", "Music", "Football", "Chess"
}

assert set(df["city"].dropna().unique()) <= {
    "Cairo", "Giza", "Alexandria"
}
```

### A3. Clean names and emails

```python
df["full_name"] = (
    df["full_name"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)

df["email"] = (
    df["email"]
    .str.strip()
    .str.lower()
)
```

```python
print(df["full_name"].value_counts(dropna=False))
print()
print(df["email"].value_counts(dropna=False))
```

### A4. Clean fee_paid

```python
df["fee_paid"].value_counts(dropna=False)
```

```python
df["fee_paid"] = (
    df["fee_paid"]
    .str.strip()
    .str.lower()
)

fee_map = {
    "yes": True,
    "y": True,
    "true": True,
    "1": True,
    "no": False,
    "n": False,
    "false": False,
    "0": False
}

df["fee_paid"] = df["fee_paid"].map(fee_map)
```

```python
df["fee_paid"].value_counts(dropna=False)
```

```python
df["fee_paid"].dtype
```

### A5. Parse signed_up_at

```python
df["signed_up_at"].value_counts(dropna=False)
```

```python
df["signed_up_at"] = pd.to_datetime(
    df["signed_up_at"],
    format="mixed",
    dayfirst=True
)
```

```python
df["signed_up_at"].head()
```

```python
df["signed_up_at"].dtype
```

```python
assert pd.api.types.is_datetime64_any_dtype(df["signed_up_at"])
assert df["signed_up_at"].notna().all()
```

## Part B — Duplicates

### B1. Remove exact duplicates

```python
exact_duplicates = df.duplicated().sum()
print("Exact duplicates:", exact_duplicates)
```

```python
before_exact = len(df)

df = df.drop_duplicates()

after_exact = len(df)

print("Rows before:", before_exact)
print("Exact duplicates removed:", before_exact - after_exact)
print("Rows after:", after_exact)
```

### B2. Keep the latest submission per student and club

```python
repeated_signups = df.duplicated(
    subset=["student_id", "club"],
    keep=False
)

df[repeated_signups].sort_values(
    ["student_id", "club", "signed_up_at"]
)
```

```python
before_repeat = len(df)

df = (
    df.sort_values("signed_up_at")
      .drop_duplicates(
          subset=["student_id", "club"],
          keep="last"
      )
)

after_repeat = len(df)

print("Rows before:", before_repeat)
print("Repeated sign-ups removed:", before_repeat - after_repeat)
print("Rows after:", after_repeat)
```

### B3. Verify one row per student per club

```python
assert not df.duplicated(
    subset=["student_id", "club"]
).any()
```

```python
print("Final rows:", len(df))
```

```python
df.to_csv("club_signups_clean.csv", index=False)
```
